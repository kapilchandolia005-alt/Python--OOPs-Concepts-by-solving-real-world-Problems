# 1. Classes and Objects

## Task 1: Rectangle Class

Create a `Rectangle` class with:
- Length
- Width
- `calculate_area()`
- `calculate_perimeter()`

```python
# Rectangle class
class Rectangle:
    def __init__(self, length, width):
        # Initialize length and width
        self.length = length
        self.width = width

    def calculate_area(self):
        # Calculate and return area
        return self.length * self.width

    def calculate_perimeter(self):
        # Calculate and return perimeter
        return 2 * (self.length + self.width)


# Taking input from the user
length = float(input("Enter rectangle length: "))
width = float(input("Enter rectangle width: "))

# Create Rectangle object
rectangle = Rectangle(length, width)

# Display results
print("\nRectangle Details")
print("Length:", rectangle.length)
print("Width:", rectangle.width)
print("Area:", rectangle.calculate_area())
print("Perimeter:", rectangle.calculate_perimeter())
```

## Task 3: BankAccount Class

Create a `BankAccount` class with:
- Account number
- Account holder's name
- Balance
- `deposit()`
- `withdraw()`
- `get_account_details()`

```python
# BankAccount class
class BankAccount:
    def __init__(self, acc_number, acc_holder, balance):
        # Initialize account details
        self.acc_number = acc_number
        self.acc_holder = acc_holder
        self.balance = balance

    def deposit(self, amount):
        # Deposit money into the account
        if amount > 0:
            self.balance += amount
            print(f"₹{amount:.2f} deposited successfully.")
        else:
            print("Deposit amount must be greater than zero.")

    def withdraw(self, amount):
        # Withdraw money if sufficient balance is available
        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
        elif amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance -= amount
            print(f"₹{amount:.2f} withdrawn successfully.")

    def get_account_details(self):
        # Display account details
        print("Account Number:", self.acc_number)
        print("Account Holder:", self.acc_holder)
        print(f"Balance: ₹{self.balance:.2f}")


# Taking input from the user
acc_number = input("Enter account number: ")
acc_holder = input("Enter account holder name: ")
balance = float(input("Enter opening balance: "))

# Create BankAccount object
account = BankAccount(acc_number, acc_holder, balance)

print("\nInitial Account Details")
account.get_account_details()

# Test deposit and withdrawal
deposit_amount = float(input("\nEnter amount to deposit: "))
account.deposit(deposit_amount)

withdraw_amount = float(input("Enter amount to withdraw: "))
account.withdraw(withdraw_amount)

print("\nFinal Account Details")
account.get_account_details()
```

# 2. Inheritance and Polymorphism

## Task 4: Shape, Circle, Rectangle and Triangle

The `Shape` class acts as a base class. The derived classes override `calculate_area()` to calculate the area of their respective shapes.

```python
# Base class
class Shape:
    def calculate_area(self):
        # This method is intended to be overridden
        raise NotImplementedError("Derived classes must implement calculate_area().")


# Derived class: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14159 * self.radius ** 2


# Derived class: Rectangle
class RectangleShape(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width


# Derived class: Triangle
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def calculate_area(self):
        return 0.5 * self.base * self.height


# Create objects
circle = Circle(5)
rectangle_shape = RectangleShape(10, 4)
triangle = Triangle(8, 6)

# Polymorphism: same method name, different implementations
shapes = [circle, rectangle_shape, triangle]

print("Shape Areas")
print("Circle Area:", circle.calculate_area())
print("Rectangle Area:", rectangle_shape.calculate_area())
print("Triangle Area:", triangle.calculate_area())
```

## Task 5: Vehicle, Car, Motorcycle and Bus

The `Vehicle` class is the base class. Each derived class overrides `start_engine()` with a vehicle-specific message.

```python
# Base class
class Vehicle:
    def start_engine(self):
        # This method is intended to be overridden
        raise NotImplementedError("Derived classes must implement start_engine().")


# Derived class: Car
class Car(Vehicle):
    def start_engine(self):
        print("Car engine has started.")


# Derived class: Motorcycle
class Motorcycle(Vehicle):
    def start_engine(self):
        print("Motorcycle engine has started.")


# Derived class: Bus
class Bus(Vehicle):
    def start_engine(self):
        print("Bus engine has started.")


# Create objects and test polymorphism
vehicles = [Car(), Motorcycle(), Bus()]

print("Vehicle Engine Status")
for vehicle in vehicles:
    vehicle.start_engine()
```

## Task 6: Animal, Dog, Cat and Lion

The `Animal` class is the base class. Each derived class overrides `make_sound()` to print the appropriate animal sound.

```python
# Base class
class Animal:
    def make_sound(self):
        # This method is intended to be overridden
        raise NotImplementedError("Derived classes must implement make_sound().")


# Derived classes
class Dog(Animal):
    def make_sound(self):
        print("Dog says: Woof!")


class Cat(Animal):
    def make_sound(self):
        print("Cat says: Meow!")


class Lion(Animal):
    def make_sound(self):
        print("Lion says: Roar!")


# Create objects and test polymorphism
animals = [Dog(), Cat(), Lion()]

print("Animal Sounds")
for animal in animals:
    animal.make_sound()
```

## Task 9: Product Class

Create a `Product` class with:
- Product name
- Price
- Quantity
- `calculate_total_cost()`

The total cost is calculated as:

**Total Cost = Price × Quantity**

```python
# Product class
class Product:
    def __init__(self, name, price, quantity):
        # Initialize product details
        self.name = name
        self.price = price
        self.quantity = quantity

    def calculate_total_cost(self):
        # Calculate and return total cost
        return self.price * self.quantity


# Taking input from the user
name = input("Enter product name: ")
price = float(input("Enter product price: "))
quantity = int(input("Enter product quantity: "))

# Create Product object
product = Product(name, price, quantity)

# Display product details
print("\nProduct Details")
print("Product Name:", product.name)
print(f"Price: ₹{product.price:.2f}")
print("Quantity:", product.quantity)
print(f"Total Cost: ₹{product.calculate_total_cost():.2f}")
```
